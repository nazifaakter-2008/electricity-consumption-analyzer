```python
appliances = [
    ["Fan", 75, 8],
    ["LED Light", 10, 5],
    ["Laptop", 100, 6]
]

total_energy = 0
for appliance in appliances:
  name = appliance[0]
  power = appliance[1]
  hours = appliance[2]

  energy = (power * hours)/1000
  total_energy = total_energy + energy

  print(name,"uses", energy, "kwh per day")

print("Total daily energy:", total_energy, "kwh")
print("Monthly energy:", total_energy*30, "kwh")
```

```python
import matplotlib.pyplot as plt
names = ["Fan","LED Light", "Laptop"]
energies = [0.6, 0.05, 0.6]

plt.bar(names, energies)
plt.title("Daily Electricity Consumption")
plt.xlabel("Appliance")
plt.ylabel("Energy (kwh)")
plt.show()
```
